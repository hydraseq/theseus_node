# Isolate near axis

```python
# This setting captures stop words
f1 = 'tests/data/poe/poe_cask_of_amontillado.txt'
f2 = 'tests/data/poe/poe_raven.txt'
x, y, words = cp.generate_xy_data(f1, f2, 0.9, True)
gr.plot_words_xy(x, y, words, n=15)
```
